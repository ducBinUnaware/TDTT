```python
#1
def merge(nums1 : list, nums2 : list):
    i=0
    j=0
    res=[]
    while(i<len(nums1) and j<len(nums2)):
        if (nums1[i]< nums2[j]):
            res.append(nums1[i])
            i+=1
        else:
            res.append(nums2[j])
            j+=1
    while(i<len(nums1)):
        res.append(nums1[i])
        i+=1
    while(j<len(nums2)):
        res.append(nums2[j])
        j+=1
    return res
a=list(map(int,input().split()))
b=list(map(int,input().split()))
print(merge(a,b))
```

```python
#2
from bisect import bisect_right
n,m=map(int,input().split())
nums=[list(map(int,input().split())) for _ in range(n)]
k=int(input())
def fin(x):
    count=0
    for i in range(n):
        count+=bisect_right(nums[i],x)
    return count>=k
l=0
r=nums[n-1][m-1]
while l<r:
    mid=(l+r)//2
    if fin(mid):
        r=mid
    else:
        l=mid+1
print(r)
```

```python
#3
def swap(s):
    nums=[]
    for i in s:
        nums.append(i)
    min_pos=len(s)-1
    res=int(s)
    for i in range(len(s)-1,-1,-1):
        nums[i],nums[min_pos]=nums[min_pos],nums[i]
        k=str()
        for j in nums:k+=j
        res=(min(res,int(k)) if (i!=len(s)-2) else int(k))
        nums[i],nums[min_pos]=nums[min_pos],nums[i]
        if nums[i]<nums[min_pos]:
            min_pos=i
    return res
s=input()
print(swap(s))
```

```python
#4
def fstr(s1,s2):
    l=0
    for i in s2:
        x=False
        while l<len(s1):
            if i==s1[l] or i==chr(ord(s1[l])-32):
                x=True
                l+=1
                break
            l+=1
        if not x: return x
    return True
n=int(input())
for i in range(n):
    s1=input()
    s2=input()
    print("yes")if fstr(s1,s2) else print("no")
```

```python
#5
def e(l1,l2):
    for i in l1:
        if not i in l2:
            return False
    return True
l1=list(map(int,input().split()))
l2=list(map(int,input().split()))
print("yes") if e(l1,l2) else print("no")
```

```python
#6
def perm(nums):
    n=len(nums)
    i=n-1
    while(i>0):
        if (nums[i]>nums[i-1]): 
            i-=1
            break
        i-=1
    if i==n-1: return reversed(nums)
    min_pos=i+1
    minn=n+1
    for j in range(n-1,i,-1):
        if nums[j]<minn and nums[j]>nums[i]:
            min_pos=j
            minn=nums[j]
    nums[i],nums[min_pos]=nums[min_pos],nums[i]
    nums[i+1:]=reversed(nums[i+1:])
    return nums
l=list(map(int,input().split()))
print(perm(l))
```

```python
#7
def canPlace(flo : list,k : int):
    n=len(flo)
    bool_flo=[False]*n
    if n>1:
        if flo[0]==0 and flo[1]==0:
            bool_flo[0]=True
        if flo[n-1]==0 and flo[n-1]==0:
            bool_flo[n-1]=True
    for i in range(1,n-1):
        if flo[i]==0 and flo[i-1]==0 and flo[i+1]==0: bool_flo[i]=True
    for i in bool_flo:
        if i: k-=1
    return k<=0
l=list(map(int,input().split()))
k=int(input())
print(canPlace(l,k))
```

```python
#8
def ispathcross(s : str):
    se=set()
    x=0
    y=0
    se.add((x,y))
    for i in s:
        if i=='R': x+=1
        elif i=='L': x-=1
        elif i=='U': y+=1
        else: y-=1
        if (x,y) in se:
            return True
        se.add((x,y))
    return False
s=input()
print(ispathcross(s))
```

```python
#9
def median(nums : list,left : int):
    if len(nums)==1:
        return nums[0]
    low=[]
    same=[]
    high=[]
    for i in nums:
        if i<nums[left]: low.append(i)
        elif i==nums[left]: same.append(i)
        else: high.append(i)
    if left< len(low):
        return median(low,left)
    elif left<len(low)+len(same):
        return same[0]
    else:
        return median(high,left-len(low)-len(same))
l=list(map(int,input().split()))
print(median(l,len(l)//2))
    
```
