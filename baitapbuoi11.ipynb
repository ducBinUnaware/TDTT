```python
#3
def happy(n,se):
    if n==1: return True
    s=str(n)
    total=0
    for i in s:
        total+=int(i)**2
    if total in se: return False
    se.add(total)
    return happy(total,se)
n=int(input())
se=set()
se.add(n)
print(happy(n,se))
```
