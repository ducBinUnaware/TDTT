```python
#9
import csv
def grade10(path : str):
    res=[]
    with open(path, 'r') as f:
        line=csv.DictReader(f)
        for row in line:
            if row["THCS4"]=="10":
                res.append(f"{row["Ho"]} {row["Ten"]}")
    return res
a=input()
f=open("output.txt",'w')
for i in grade10(a):
    f.write(i+'\n')
f.close()
```

```python
#10
def getMoney(path : str) -> list:
    e=[]
    o=[]
    sume=0
    sumo=0
    with open(path , 'r') as f:
        for line in f:
            l=line.split()
            for i in range(0,len(l),2):
                sume+=int(l[i])
                e.append(l[i])
            for i in range(1,len(l),2):
                sumo+=int(l[i])
                o.append(l[i])
    return e if sume>sumo else o
a=input()
print(getMoney(a))
```

```python
#12
def maximumPro(path : str) -> int:
    with open(path,'r' ) as f:
        line=f.readline().split()
        for i in range(len(line)): line[i]=int(line[i])
        line=sorted(line)
        if len(line)<3:
            r=1
            for i in line: r*=i
            return r
        return max(line[0]*line[1]*line[-1],line[-1]*line[-2]*line[-3])
a=input()
print(maximumPro(a))
```

```python
#13
def moveZeroes(path : str) -> list:
    res=[]
    with open(path, 'r') as f:
        n=int(f.readline())
        l=f.readline().split()
        count0=0
        for i in l:
            if i=='0': count0+=1
            else: res.append(i)
        for i in range(count0):
            res.append('0')
    return res
a=input()
print(moveZeroes(a))
```

```python
#15
import csv
def findMovies(path : str):
    res=[]
    with open(path , 'r') as f:
        line= csv.DictReader(f, skipinitialspace=True)
        for row in line:
            if float(row["rating"])>=8.0 and row["description"]!="boring":
                res.append(row["movie"])
    return res
a=input()
print(findMovies(a)) if findMovies(a) else print("All are boring!")
```

```python
#16
pi=3.14
def perimeter(path : str):
    res=[]
    with open(path , 'r') as f:
        lines=[line.strip() for line in f if line.strip()]
        i=0
        while i<len(lines):
            if "TYPE: SQUARE" in lines[i]:
                side=float(lines[i+1].split(": ")[1])
                res.append(4*side)
                i+=2
            elif "TYPE: RECTANGLE" in lines[i]:
                w=float(lines[i+1].split(": ")[1])
                h=float(lines[i+2].split(": ")[1])
                res.append((w+h)*2)
                i+=3
            elif "TYPE: CIRCLE" in lines[i]:
                r=float(lines[i+1].split(": ")[1])
                res.append(2*pi*r)
                i+=2
    return res
a=input()
for i in perimeter(a): print(i)
```

```python
#17
def productexs(path : str):
    res=[]
    with open(path, 'r') as f:
        n=int(f.readline().strip())
        l=f.readline().split()
        for i in range(n): l[i]=int(l[i])
        for i in range(n):
            temp=1
            for j in range(n):
                if j!=i: temp*=l[j]
            res.append(temp)
    return res
a=input()
print(productexs(a))
```

```python
#18
def esTime(path : str):
    with open(path, 'r') as f:
        line1=f.readline().split()
        x1=int(line1[2])
        y1=int(line1[4])
        line2=f.readline().split()
        x2=int(line2[2])
        y2=int(line2[4])
        line3=f.readline().split()
        v=int(line3[1])
        return ((abs(x1-x2)**2+abs(y1-y2)**2)**0.5)//v
a=input()
print(esTime(a))
```

```python
#19
def readBinary(path):
    res=[]
    with open(path, 'rb') as f:
        for line in f:
            l=f.readline().split()
            for i in l:
                res.append(i)
    return res
a=input()
for i in readBinary(a):
    print(f"{i:.5f}")
```

```python
#20
def images(path):
    with open(path,'r') as f:
        M=int(f.readline().strip())
        img=[]
        for i in range(M):
            line=list(map(int,f.readline().split()))
            img.append(line)
        res=[[0]*M for i in range(M)]
        for i in range(M):
            for j in range(M):
                sum=0
                cur=0
                for k in range(i-1,i+2):
                    for t in range(j-1,j+2):
                        if 0<=k<M and 0<=t<M:
                            sum+=img[k][t]
                            cur+=1
                res[i][j]=sum//cur
        for i in res:
            print(*i)
a=input()
images(a)
```
